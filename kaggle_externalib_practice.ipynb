# Now lets toss a dice 10 times using np.random built-in function : randint

```python
DiceTopNumbers = np.random.randint(low = 1 , high = 6 , size = 20 )
print(DiceTopNumbers)
print ('What is DiceTopNumbers?' , type(DiceTopNumbers));
print ('What RandInt submodule can do?' , dir(DiceTopNumbers));
print('Average of dice numbers: ', DiceTopNumbers.mean());
DiceTopNumbers.tolist();
print('List of Dice tossing results:',DiceTopNumbers);
print('Type of DiceTopNumbers is = ' , type(DiceTopNumbers.tolist()) );
```

# Now lets overload (positive) operator to make it act as an appending function, so wen you add <int> to a <list> , it appends <int> to <list>

```python
rolls = np.random.randint(low = 1 , high = 6 , size = 10)
print(rolls)
print(rolls+3);
print(rolls<=3);
```

Now lets convert matrix to an nd(N-Domensional) array so we can deal with multi-dimensional data & to access elements of array with indices

```python
m = [[1,2,3] , [2,4,6]];
print('first element in M list is: ' , m[0]);
print('type of my matrix:', type(m));
m1 = np.asarray(m);
print('type of my matrix:', type(m1));
print ('last element of first record is ' , m1[0,-1]);
```
